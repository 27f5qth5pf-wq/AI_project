```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df1 = pd.read_csv(r"C:\Users\HP\Desktop\AI_project\week-04-libraries-descriptive-statistics\datasets\students.csv")
df1['gender']= df1['gender'].str.strip().str.title()
df1['city']= df1['city'].str.strip().str.title()

for col in ['study_hours', 'attendance' , 'prev_gpa']:
    df1[col] = df1[col].fillna(df1[col].median())
df1 = df1.drop_duplicates()

df1.head()
```
