```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

x = np.random.randint(1, 20, size=5)
y = np.random.randint(50, 120, size=5)

df = pd.DataFrame({'x': x, 'y': y})

coeffs = np.polyfit(x, y, 1)
y_pred = np.polyval(coeffs, x)

print('x:', x)
print('y:', y)
print('linear fit coeffs (slope, intercept):', coeffs)

plt.scatter(df['x'], df['y'], label='random points')
x_sorted = np.sort(df['x'])
plt.plot(x_sorted, np.polyval(coeffs, x_sorted), color='red', label='linear fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

## model

```python
def predict( x, m ,b ):
    y_hat = m * x + b
    return  y_hat
```

# Cost function

```python
def cost(y ,y_hat):
    c= np.sum ((y - y_hat)**2)/(2*len(y))
    # c = (1/2(2*len(y))) * np.sum((y - y_hat)**2)
    return c
```

# Gradients

```python
def gradients(x ,y , y_hat):
    gm = np.sum(( y - y_hat ) * x )
    gb = np.sum(y_hat - y) / len(x)

    return gm , gb
```

# show

```python
def show(x , y ,m , b):
    y_hat = predict(x , m, b)
    plt.scatter(x, y , s= 60 , marker = "x" , label = "Actual data")
    plt.plot(x , y_hat , label = f"line for m {m} , b {b}")
    plt.xlabel("Hours studied")
    plt.ylabel("Score")
    plt.grid()
    plt.legend()
    plt.show()
```

```python
m , b = 0 , 0

y_hat = predict(x ,m ,b)
print("Actual y : ",  y)
print("model frediction : " , y_hat)

loss = cost(y , y_hat)
print( f"when = {m} and b = {b} the cost = {loss} ")
```

```python
show(x,y,m,b)
```
