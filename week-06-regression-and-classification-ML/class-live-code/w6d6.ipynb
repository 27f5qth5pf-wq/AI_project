```python
import numpy as np

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([0, 0, 0, 1, 1], dtype=float)

w = 0.0
b = 0.0

learning_rate = 0.1
epochs = 1000

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

m = len(X)

for epoch in range(epochs):
    z = w * X + b
    y_pred = sigmoid(z)

    loss = -(1 / m) * np.sum(
        y * np.log(y_pred + 1e-9) +
        (1 - y) * np.log(1 - y_pred + 1e-9)
    )

    dw = (1 / m) * np.sum((y_pred - y) * X)
    db = (1 / m) * np.sum(y_pred - y)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    if epoch % 100 == 0:
        print(f"Epoch: {epoch:4d}  Loss: {loss:.4f}")

print("\nWeight:", w)
print("Bias:", b)

hours = 3.5

z = w * hours + b
probability = sigmoid(z)

print("\nProbability:", probability)

if probability >= 0.5:
    print("Prediction: PASS")
else:
    print("Prediction: FAIL")
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([0, 0, 0, 1, 1], dtype=float)

w = 0.0
b = 0.0

learning_rate = 0.1
epochs = 1000

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

m = len(X)
loss_history = []

for epoch in range(epochs):
    z = w * X + b
    y_pred = sigmoid(z)

    loss = -(1 / m) * np.sum(
        y * np.log(y_pred + 1e-9) +
        (1 - y) * np.log(1 - y_pred + 1e-9)
    )

    loss_history.append(loss)

    dw = (1 / m) * np.sum((y_pred - y) * X)
    db = (1 / m) * np.sum(y_pred - y)

    w -= learning_rate * dw
    b -= learning_rate * db

    if epoch % 100 == 0:
        print(f"Epoch: {epoch:4d}  Loss: {loss:.4f}")

print("\nWeight:", w)
print("Bias:", b)

hours = 3.5
probability = sigmoid(w * hours + b)

print("Probability:", probability)

if probability >= 0.5:
    print("Prediction: PASS")
else:
    print("Prediction: FAIL")

# -----------------------------
# Graph 1: Loss vs Epoch
# -----------------------------
plt.figure(figsize=(8, 5))
plt.plot(loss_history)
plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.grid(True)
plt.show()

# -----------------------------
# Graph 2: Sigmoid Curve + Data
# -----------------------------
x_plot = np.linspace(0, 6, 200)
y_curve = sigmoid(w * x_plot + b)

plt.figure(figsize=(8, 5))
plt.plot(x_plot, y_curve, label="Sigmoid Curve")
plt.scatter(X, y, s=80, label="Training Data")
plt.axhline(0.5, linestyle="--", label="Decision Boundary")
plt.title("Logistic Regression")
plt.xlabel("Hours Studied")
plt.ylabel("Probability")
plt.xlim(0, 6)
plt.ylim(-0.1, 1.1)
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([
    [25, 30000],
    [30, 50000],
    [35, 60000],
    [40, 80000]
], dtype=float)

y = np.array([0, 0, 1, 1])

w = np.zeros(X.shape[1])
b = 0

lr = 0.000001
epochs = 1000

losses = []

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

for epoch in range(epochs):

    z = np.dot(X, w) + b
    y_pred = sigmoid(z)

    loss = -np.mean(
        y * np.log(y_pred + 1e-9) +
        (1 - y) * np.log(1 - y_pred + 1e-9)
    )

    losses.append(loss)

    dw = np.dot(X.T, (y_pred - y)) / len(X)
    db = np.mean(y_pred - y)

    w -= lr * dw
    b -= lr * db

new_data = np.array([[32, 55000]])

probability = sigmoid(np.dot(new_data, w) + b)
prediction = (probability >= 0.5).astype(int)

print("Weight:", w)
print("Bias:", b)
print("Probability:", probability)
print("Prediction:", prediction)

plt.plot(losses)
plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```
